```python
import pandas as pd
import ast
import numpy as np
```

```python
df = pd.read_csv('HPXMLinputs.csv', sep=',', index_col=False)
```

```python
listAttributes = df[df['Type'].isin(['Choice','Boolean'])]['Name'].tolist()
```

```python
data_types = {'Boolean': bool,
			  'Integer': int,
			  'Choice': str,
			  'Double': float,
			  'String': str}
```

```python
listNames = df['Name'].tolist()
df_2 = df[['Type','Required','Choices','Default Value']].to_dict(orient='records')
new_dict = dict(zip(listNames, df_2))
```

```python
for i in new_dict.keys():
    if pd.isna(new_dict[i]['Default Value']) == False:
        new_dict[i]['Default Value'] = data_types[new_dict[i]['Type']](new_dict[i]['Default Value'])
```

```python
for i in listAttributes:
	new_dict[i]['Choices'] = ast.literal_eval(new_dict[i]['Choices'])
```

```python
#dump the dictionary to a file
import json
with open('hpxml-args-constraints.json', 'w') as f:
	json.dump(new_dict, f, indent=4)
```
